The use of spacy for tokenization.

```python
# import required libraries
import spacy
from nltk.tokenize import TweetTokenizer
```

```python
# Load the specific model
nlp = spacy.load("en_core_web_sm")
```

```python
# input text
text = "Mary don't slap the green witch."
```

```python
# input tweet 
tweet = u"Snow White and the Seven Degrees #MakeAMovieCold@midnight:-)"
```

```python
print([str(token) for token in nlp(text.lower())])
```

```python
print([str(token) for token in nlp(tweet.lower())])
```

The use of nltk tokenizer.

```python
# initialize a tokenizer object 
tokenizer = TweetTokenizer()
```

```python
print(tokenizer.tokenize(text.lower()))
```

```python
print(tokenizer.tokenize(tweet.lower()))
```

See the difference in the output tokens using by spacy and nltk TweetTokenizer. Both tokenizer performs well on the "Mary don't slap the green witch." sentence. On the other hand, TweetTokenizer from nltk peforms well on hastag and emoji but spacy fails on creating token from these aspects. This can be because of I used the smallest tokenization model, en_core_web_sm. 

### Sample n-grams

```python
def n_grams(text, n):
    """ N-grams function """
    return [text[i:i+n] for i in range(len(text)-n+1)]
```

```python
cleaned = ['mary', ',', "n't", 'slap', 'green', 'witch', '.']
```

```python
print(n_grams(cleaned,3))
```

```python
from nltk.util import ngrams

sentence = "I love natural language processing".split()
```

```python
bigrams = list(ngrams(sentence, 2))
print(bigrams)
```

```python
trigrams = list(ngrams(sentence, 3))
print(trigrams)
```

### Lemmas & Stems 

Lemmas are root forms of words. For the verb fly, it can be inflected into many different words flow, flew, flown, flies, flowing, etc. The word fly is the lemma for all of these seemingly different words.

```python
doc = nlp(u"he was running late")
```

```python
for token in doc:
    print('{} --> {}'.format(token, token.lemma_))
```

### Part of speech Tagging (POS tagging)

```python
for token in doc:
    print('{} - {}'.format(token, token.pos_))
```

### Chunking 

```python
for chunk in doc.noun_chunks:
    print('{} - {}'.format(chunk, chunk.label_))
```
