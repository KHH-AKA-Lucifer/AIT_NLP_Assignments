The use of spacy for tokenization.

```python
# import required libraries
import spacy
from nltk.tokenize import TweetTokenizer
```

```python
# Load the specific model
nlp = spacy.load("en_core_web_sm")
```

```python
# input text
text = "Mary don't slap the green witch."
```

```python
# input tweet 
tweet = u"Snow White and the Seven Degrees #MakeAMovieCold@midnight:-)"
```

```python
print([str(token) for token in nlp(text.lower())])
```

```python
print([str(token) for token in nlp(tweet.lower())])
```

The use of nltk tokenizer.

```python
# initialize a tokenizer object 
tokenizer = TweetTokenizer()
```

```python
print(tokenizer.tokenize(text.lower()))
```

```python
print(tokenizer.tokenize(tweet.lower()))
```

See the difference in the output tokens using by spacy and nltk TweetTokenizer. Both tokenizer performs well on the "Mary don't slap the green witch." sentence. On the other hand, TweetTokenizer from nltk peforms well on hastag and emoji but spacy fails on creating token from these aspects. This can be because of I used the smallest tokenization model, en_core_web_sm.
